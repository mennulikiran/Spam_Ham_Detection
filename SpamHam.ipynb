```python
bow  = CountVectorizer(stop_words="english")
```

```python
X=df['Message']
Y=df['Category']
```

```python
final_data1 = pd.DataFrame(bow.fit_transform(X).toarray(),columns=bow.get_feature_names_out())
```

```python
final_data1
```

## Data Splitting

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(final_data1,Y,test_size=0.25,random_state=20) # random_state used to make data should be constant even the dataset executed again and again
```

```python
x_train
```

## Model Building

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier()
knn.fit(x_train, y_train)
```

```python
y_pred = knn.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```
